### **Задание 1**
Определите, сколько футболистов представляют каждый клуб, и найдите топ-5 самых популярных клубов.

```
# Группировка игроков по клубам и вывод топ-5
club_counts = data['Club'].value_counts()
print("Топ-5 клубов по количеству игроков:")
print(club_counts.head(5))
```

### **Задание 2**
Посчитайте среднюю и медианную зарплату (Wage) футболистов в зависимости от их позиции на поле (Position).

```
# Расчет средней и медианной зарплаты для каждой позиции
wage_by_position = data.groupby('Position')['Wage'].agg(['mean', 'median'])
print("Анализ зарплат по позициям:")
print(wage_by_position)
```

### **Задание 3**
Постройте гистограмму распределения возраста футболистов.

```
# Визуализация распределения возраста игроков
plt.figure(figsize=(10, 6))
plt.hist(data['Age'], bins=20, color='skyblue', edgecolor='black')
plt.title('Распределение возраста футболистов (Вариант 12)')
plt.xlabel('Возраст')
plt.ylabel('Количество игроков')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
